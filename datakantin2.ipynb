```python
import pandas as pd
import numpy as np

# Load dataset langsung dari folder Colab
df = pd.read_csv('data_kantin.csv')

# Inspeksi Awal
print("--- 5 Baris Pertama ---")
print(df.head())

print("\n--- Info Tipe Data & Missing Value ---")
print(df.info())

print("\n--- Statistik Deskriptif ---")
print(df.describe())

print("\n--- Dimensi Data (Baris, Kolom) ---")
print(df.shape)
```

```python
# Cek data kosong sebelum dibersihkan
print("Missing value awal:")
print(df.isnull().sum())

# 1. Handling Missing Value (Isi 'terjual' kosong dengan 0, hapus baris jika 'menu' kosong)
df['terjual'] = df['terjual'].fillna(0)
df = df.dropna(subset=['menu'])

# 2. Handling Duplikat
df = df.drop_duplicates()

# 3. Perbaikan Tipe Data
df['terjual'] = df['terjual'].astype(int)

print("\n--- Tipe Data Setelah Cleaning ---")
print(df.dtypes)
print("\nDimensi data setelah cleaning:", df.shape)
```

```python
# 1. Kolom Turunan: Total Pendapatan
df['total_pendapatan'] = df['harga'] * df['terjual']

# 2. Filtering: Terjual > 20
laris = df[df['terjual'] > 20]
print("--- Menu Terjual > 20 ---")
print(laris.head())

# 3. Sorting: Terjual terbanyak
urut = df.sort_values(by='terjual', ascending=False)
print("\n--- Top 5 Penjualan Terbanyak ---")
print(urut.head())

# 4. Groupby per Menu
ringkasan_menu = df.groupby('menu').agg(
    total_terjual=('terjual', 'sum'),
    rata_rata_terjual=('terjual', 'mean'),
    total_pendapatan=('total_pendapatan', 'sum')
).reset_index().sort_values(by='total_pendapatan', ascending=False)

print("\n--- Ringkasan Penjualan per Menu ---")
print(ringkasan_menu)

# 5. Groupby per Kategori
ringkasan_kategori = df.groupby('kategori').agg(
    total_pendapatan=('total_pendapatan', 'sum')
).reset_index().sort_values(by='total_pendapatan', ascending=False)

print("\n--- Ringkasan Pendapatan per Kategori ---")
print(ringkasan_kategori)
```
